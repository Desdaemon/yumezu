# yume.wiki SMW queries

```python
# preparations
%pip install requests
import requests
import json
```

- All requests have a hard limit of 500 items, any more must be retrieved from a continue key under `query-continue-offset`
- The continue key must be appended to `parameters` as `offset=$continue_key`

## Locations

```python
requests.get(
    'https://yume.wiki/api.php',
    headers={'Origin': 'https://explorer.yume.wiki'},
    params={
        'action': 'askargs',
        'format': 'json',
        'conditions': 'Category:Yume 2kki Locations',
        'printouts': 'Has location image|Header background color|Header font color|Has primary author|Has contributing author|Japanese name|Has BGM|Map IDs|Has location map|Version added|Versions updated|Version removed|Version gaps',
        'parameters': 'limit=10',
        'api_version': 3,
    }
).json()['query']['results']
```

## Connections

```python
requests.get(
    'https://yume.wiki/api.php',
    headers={'Origin': 'https://explorer.yume.wiki'},
    params={
        'action': 'askargs',
        'format': 'json',
        'conditions': 'Yume 2kki:+|Is subobject type::connection',
        'printouts': 'Connection/Origin|Connection/Location|Connection/Attribute|Connection/Unlock conditions|Connection/Effects needed|Connection/Season available|Connection/Chance percentage|Connection/Chance description|Connection/Is removed',
        'parameters': 'limit=10',
        'api_version': 3,
    }
).json()['query']['results']
```

## Artists

```python
requests.get(
    'https://yume.wiki/api.php',
    headers={'Origin': 'https://explorer.yume.wiki'},
    params={
        'action': 'askargs',
        'format': 'json',
        'conditions': '-Has subobject::Yume 2kki:Authors',
        'printouts': 'Author/Name|Author/Original Name',
        'parameters': 'limit=10|sort=Author/Name|order=asc',
        'api_version': 3,
    }
).json()['query']['results']
```

## Versions (+patches)

```python
requests.get(
    'https://yume.wiki/api.php',
    headers={'Origin': 'https://explorer.yume.wiki'},
    params={
        'action': 'askargs',
        'format': 'json',
        'conditions': 'Is part of game::Yume 2kki|Version/Type::+',
        'printouts': 'Version|Version/Type|Has contributing author|Version/Date',
        'parameters': 'limit=10|sort=Version/Date|order=desc',
        'api_version': 3,
    }
).json()['query']['results']
```
